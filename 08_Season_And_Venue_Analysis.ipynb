### Notebook Objective
1. Track the macro scoring evolution of the IPL across 19 editions (total runs, wickets, run rate).
2. Analyze stadium dynamics across top IPL venues: matches hosted, average 1st innings scores, and boundary frequency.
3. Assess venue chasing bias (defending vs chasing win percentages).


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

clean_dir = os.path.join(PROJECT_DIR, 'Cleaned_Data')
pbi_dir = os.path.join(PROJECT_DIR, 'PowerBI')
viz_dir = os.path.join(PROJECT_DIR, 'Visualizations')

matches = pd.read_csv(os.path.join(clean_dir, 'matches.csv'))
deliveries = pd.read_csv(os.path.join(clean_dir, 'deliveries.csv'), low_memory=False)
season_stats = pd.read_csv(os.path.join(pbi_dir, 'Fact_Season_Summary.csv'))
```

### 1. Season Scoring Evolution


```python
season_stats[['season', 'matches', 'total_runs', 'total_wickets', 'avg_match_runs', 'overall_run_rate', 'total_fours', 'total_sixes']]
```

```python
fig, ax1 = plt.subplots(figsize=(10, 5))
ax2 = ax1.twinx()
ax1.bar(season_stats['season'], season_stats['total_runs'], color='#2ca02c', alpha=0.7, label='Total Runs')
ax2.plot(season_stats['season'], season_stats['overall_run_rate'], color='#1f77b4', marker='o', linewidth=2.5, label='Run Rate (RPO)')
ax1.set_title("IPL Scoring Progression: Total Runs & Run Rate Across Editions", fontsize=13, fontweight='bold')
ax1.set_xlabel("Season")
ax1.set_ylabel("Total Runs Scored")
ax2.set_ylabel("Overall Run Rate (RPO)")
ax1.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.savefig(os.path.join(viz_dir, '10_season_scoring_trends.png'), dpi=200)
plt.show()
```

### 2. Venue Scoring Patterns


```python
top_venues = matches['venue'].value_counts().head(10).index
venue_df = deliveries[deliveries['venue'].isin(top_venues) & (deliveries['innings'] == 1) & ~deliveries['is_super_over']]
venue_scores = venue_df.groupby(['venue', 'match_id'])['total_runs'].sum().groupby('venue').mean().sort_values(ascending=True)

plt.figure(figsize=(10, 6))
bars = plt.barh(venue_scores.index, venue_scores.values, color='#9467bd')
plt.title("Average 1st Innings Score Across Top 10 IPL Venues", fontsize=13, fontweight='bold')
plt.xlabel("Average 1st Innings Total")
plt.ylabel("Venue")
for b in bars:
    w = b.get_width()
    plt.annotate(f"{w:.1f}", xy=(w, b.get_y() + b.get_height()/2), xytext=(4, 0), textcoords="offset points", va='center')
plt.tight_layout()
plt.savefig(os.path.join(viz_dir, '11_venue_scoring_patterns.png'), dpi=200)
plt.show()
```

### 3. Venue & Season Insights
- IPL run rates have steadily risen from ~8.20 RPO in the initial seasons (2008–2010) to over 9.50+ RPO in recent editions.
- **M Chinnaswamy Stadium and Wankhede Stadium** feature the highest average 1st innings totals (~172–175), characterized by shorter boundaries and true batting surfaces.
- **MA Chidambaram Stadium (Chepauk)** favors spin and lower scoring totals (~155–160 avg), making defending totals more viable.
