```python
import os
import pandas as pd
import numpy as np

# Verify project paths
data_dir = os.path.join(PROJECT_DIR, 'Dataset')
clean_dir = os.path.join(PROJECT_DIR, 'Cleaned_Data')

print("Dataset directory:", data_dir)
print("Cleaned Data directory:", clean_dir)
```

### 1. Loading the Raw Datasets
We load the raw match information and delivery records provided by Cricsheet.


```python
# Load the cleaned matches and deliveries data
matches = pd.read_csv(os.path.join(clean_dir, 'matches.csv'))
deliveries = pd.read_csv(os.path.join(clean_dir, 'deliveries.csv'), low_memory=False)

print(f"Matches Shape: {matches.shape}")
print(f"Deliveries Shape: {deliveries.shape}")
```

### 2. Inspecting the Matches Dataset
Examining the structure of match-level metadata.

```python
# Match metadata columns and data types
matches.info()
```

```python
# Match numerical summary statistics
matches.describe()
```

### 3. Inspecting the Deliveries Dataset
Examining ball-by-ball granularity, scoring attributes, and dismissals.


```python
# Inspect first 5 ball-by-ball delivery records
deliveries.head()
```

```python
# Deliveries columns and data types
deliveries.info()
```

```python
# Summary statistics of ball-by-ball scores and extras
deliveries[['runs_off_bat', 'extras', 'wides', 'noballs', 'byes', 'legbyes', 'total_runs']].describe()
```

### 4. Tournament Scope and Overview Statistics
Summarizing the temporal span, participating franchises, and total volume of matches.


```python
total_matches = len(matches)
total_deliveries = len(deliveries)
seasons_count = matches['season'].nunique()
unique_venues = matches['venue'].nunique()
unique_batters = deliveries['striker'].nunique()
unique_bowlers = deliveries['bowler'].nunique()

print("="*60)
print("IPL DATASET COMPREHENSIVE OVERVIEW (CRICSHEET OFFICIAL)")
print("="*60)
print(f"Total Matches Played    : {total_matches:,}")
print(f"Total Deliveries Bowled : {total_deliveries:,}")
print(f"Total IPL Seasons       : {seasons_count} editions")
print(f"Total Stadiums / Venues : {unique_venues}")
print(f"Unique Batters          : {unique_batters}")
print(f"Unique Bowlers          : {unique_bowlers}")
print("="*60)
```
