### Notebook Objective
1. Identify elite dual-threat players (All-Rounders) in IPL history (Criteria: min 500 runs and min 20 wickets).
2. Evaluate all-rounders combining batting strike rate with bowling economy rate.
3. Compare match impact across top batters, bowlers, and all-rounders.


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pbi_dir = os.path.join(PROJECT_DIR, 'PowerBI')
viz_dir = os.path.join(PROJECT_DIR, 'Visualizations')

batting_stats = pd.read_csv(os.path.join(pbi_dir, 'Fact_Batting.csv'))
bowling_stats = pd.read_csv(os.path.join(pbi_dir, 'Fact_Bowling.csv'))
```

### 1. Merging Batting and Bowling Profiles


```python
all_rounders = batting_stats.merge(bowling_stats, on='player', suffixes=('_bat', '_bowl'))
all_rounders = all_rounders[(all_rounders['runs'] >= 500) & (all_rounders['wickets'] >= 20)].copy()
all_rounders = all_rounders.sort_values(by=['runs', 'wickets'], ascending=[False, False]).reset_index(drop=True)

print(f"Qualified All-Rounders: {len(all_rounders)}")
all_rounders[['player', 'runs', 'batting_average', 'strike_rate', 'wickets', 'economy_rate', 'bowling_average']].head(15)
```

### 2. All-Rounder Impact Scatter Plot
Evaluating Batting Strike Rate against Bowling Economy Rate.


```python
plt.figure(figsize=(9, 6))
top_ar = all_rounders.head(20)
scatter = plt.scatter(top_ar['strike_rate'], top_ar['economy_rate'], s=top_ar['wickets']*5, c=top_ar['runs'], cmap='plasma', alpha=0.85)
plt.title("All-Rounder Impact: Batting Strike Rate vs Bowling Economy Rate", fontsize=13, fontweight='bold')
plt.xlabel("Batting Strike Rate")
plt.ylabel("Bowling Economy Rate (Lower is Better)")

for _, r in top_ar.head(10).iterrows():
    plt.annotate(r['player'], xy=(r['strike_rate'], r['economy_rate']), xytext=(5, 5), textcoords='offset points', fontsize=9)

plt.colorbar(scatter, label='Total Career Runs')
plt.tight_layout()
plt.savefig(os.path.join(viz_dir, '09_all_rounders_performance.png'), dpi=200)
plt.show()
```

### 3. All-Rounder Insights
- **Andre Russell** stands out with the highest batting strike rate (>170) while contributing heavily with the ball in death overs.
- **Ravindra Jadeja and Dwayne Bravo** demonstrate sustained longevity with 2,500+ runs and 150+ wickets.
- **Sunil Narine** presents a unique archetype: transformed from an elite economy bowler into an explosive opening batter.
