### Notebook Objective
1. Finalize the dimensional modeling architecture (Star Schema) for the Power BI interactive dashboard.
2. Verify all dimension and fact tables:
   - `Dim_Matches`: Match metadata, venue, toss, outcome.
   - `Fact_Batting`: Career batting performance.
   - `Fact_Bowling`: Career bowling performance.
   - `Fact_Team_Performance`: Franchise records and win percentages.
   - `Fact_Season_Summary`: Season macro scoring and run rates.
   - `Fact_Match_Phase`: Innings phase metrics.
   - `Fact_Player_AllRounder`: Dual-threat all-rounder profiles.
3. Validate CSV exports in `PowerBI/` and the master multi-sheet workbook `IPL_PowerBI_Complete_Data.xlsx`.
4. Document the DAX measures catalog and 6-page Power BI dashboard design.


```python
import os
import pandas as pd
import numpy as np

pbi_dir = os.path.join(PROJECT_DIR, 'PowerBI')
clean_dir = os.path.join(PROJECT_DIR, 'Cleaned_Data')

# Load and verify all exported Power BI tables
dim_matches = pd.read_csv(os.path.join(pbi_dir, 'Dim_Matches.csv'))
fact_batting = pd.read_csv(os.path.join(pbi_dir, 'Fact_Batting.csv'))
fact_bowling = pd.read_csv(os.path.join(pbi_dir, 'Fact_Bowling.csv'))
fact_team = pd.read_csv(os.path.join(pbi_dir, 'Fact_Team_Performance.csv'))
fact_season = pd.read_csv(os.path.join(pbi_dir, 'Fact_Season_Summary.csv'))
fact_phase = pd.read_csv(os.path.join(pbi_dir, 'Fact_Match_Phase.csv'))
fact_ar = pd.read_csv(os.path.join(pbi_dir, 'Fact_Player_AllRounder.csv'))

print("All Power BI tables verified successfully!")
```

### 1. Star Schema Table Dimensions & Row Counts


```python
pbi_manifest = pd.DataFrame([
    {'Table Name': 'Dim_Matches', 'Type': 'Dimension', 'Rows': len(dim_matches), 'Columns': dim_matches.shape[1], 'Primary Key': 'match_id'},
    {'Table Name': 'Fact_Batting', 'Type': 'Fact', 'Rows': len(fact_batting), 'Columns': fact_batting.shape[1], 'Key': 'player'},
    {'Table Name': 'Fact_Bowling', 'Type': 'Fact', 'Rows': len(fact_bowling), 'Columns': fact_bowling.shape[1], 'Key': 'player'},
    {'Table Name': 'Fact_Team_Performance', 'Type': 'Fact', 'Rows': len(fact_team), 'Columns': fact_team.shape[1], 'Key': 'team'},
    {'Table Name': 'Fact_Season_Summary', 'Type': 'Fact', 'Rows': len(fact_season), 'Columns': fact_season.shape[1], 'Key': 'season'},
    {'Table Name': 'Fact_Match_Phase', 'Type': 'Fact', 'Rows': len(fact_phase), 'Columns': fact_phase.shape[1], 'Key': 'season, phase'},
    {'Table Name': 'Fact_Player_AllRounder', 'Type': 'Fact', 'Rows': len(fact_ar), 'Columns': fact_ar.shape[1], 'Key': 'player'}
])
pbi_manifest
```

### 4. Power BI Dashboard Architecture (6 Pages)
1. **Overall IPL Dashboard:** High-level executive KPIs, season scoring bars, franchise wins leaderboard.
2. **Batting Performance:** Top 10 run scorers, batting avg vs strike rate scatter plot, boundaries breakdown.
3. **Bowling Performance:** Top 10 wicket takers, economy vs strike rate scatter plot, dot ball %.

6. **Season and Match Analysis:** Evolutionary scoring trends, venue 1st innings benchmarks, toss decisions.
