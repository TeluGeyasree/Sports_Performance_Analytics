### Notebook Objective
1. Measure the correlation between winning the toss and winning the match.
2. Analyze historical toss decisions: Batting First vs Fielding First across seasons.
3. Compare match win percentage based on toss decision.
4. Identify franchises that capitalize most effectively on toss victories.


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

clean_dir = os.path.join(PROJECT_DIR, 'Cleaned_Data')
viz_dir = os.path.join(PROJECT_DIR, 'Visualizations')

matches = pd.read_csv(os.path.join(clean_dir, 'matches.csv'))
```

### 1. Overall Toss Advantage


```python
valid_toss = matches[matches['toss_winner'].notna() & matches['winner'].notna()].copy()
valid_toss['toss_winner_won'] = valid_toss['toss_winner'] == valid_toss['winner']

total_decisive = len(valid_toss)
toss_match_wins = valid_toss['toss_winner_won'].sum()
toss_win_pct = round((toss_match_wins / total_decisive) * 100, 2)

print("="*55)
print("TOSS ADVANTAGE IN IPL MATCHES")
print("="*55)
print(f"Decisive Matches Evaluated : {total_decisive:,}")
print(f"Matches Won by Toss Winner : {toss_match_wins:,}")
print(f"Toss Winner Win Percentage : {toss_win_pct}%")
print("="*55)
```

### 2. Evolution of Toss Decisions (Bat vs Field)


```python
toss_dec_season = pd.crosstab(matches['season'], matches['toss_decision'], normalize='index') * 100
toss_dec_season.round(1)
```

### 3. Match Win % by Toss Decision


```python
toss_res = valid_toss.groupby('toss_decision').agg(matches=('match_id', 'count'), wins=('toss_winner_won', 'sum')).reset_index()
toss_res['win_pct'] = ((toss_res['wins'] / toss_res['matches']) * 100).round(1)

plt.figure(figsize=(6, 4.5))
bars = plt.bar(toss_res['toss_decision'].str.upper(), toss_res['win_pct'], color=['#1f77b4', '#2ca02c'], width=0.4)
plt.title("Match Win % by Toss Decision", fontsize=12, fontweight='bold')
plt.xlabel("Toss Decision")
plt.ylabel("Win % for Toss Winner")
plt.ylim(0, 70)
for b in bars:
    h = b.get_height()
    plt.annotate(f"{h:.1f}%", xy=(b.get_x() + b.get_width()/2, h), xytext=(0, 4), textcoords="offset points", ha='center')
plt.tight_layout()
plt.savefig(os.path.join(viz_dir, '12_toss_decision_win_percentage.png'), dpi=200)
plt.show()
```

### 4. Toss Insights
- The overall toss advantage across all 19 editions is moderate (~51.5% win rate for toss winners), showing that team quality outweighs luck.
- Modern captains overwhelmingly elect to **field first** (~70–75% of tosses in recent seasons) driven by dew factors and enhanced chasing analytics.
